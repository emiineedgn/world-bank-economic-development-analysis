# World Bank Indicators – Exploratory Data Analysis

Bu çalışmada farklı ülkelerin ekonomik, sosyal ve çevresel göstergeleri Python kullanılarak incelenmiştir. Amaç, veri setindeki değişkenleri önce genel olarak tanımak, ardından yıllara ve ülkelere göre değişimleri görselleştirerek değişkenler arasındaki ilişkileri incelemektir.

Çalışmada özellikle **GDP, nüfus, yaşam beklentisi, işsizlik, kişi başına CO₂ emisyonu ve elektriğe erişim** göstergelerine odaklanılmıştır.

> **Not:** Veri seti Kaggle'da paylaşılmış, World Bank göstergelerini örnekleyen bir veri setidir. Kaynak ve veri setinin kapsamı aşağıdaki bölümde belirtilmiştir.


## 1. Veri Seti Hakkında

**Veri seti:** World Bank Dataset  
**Kaynak:** Kaggle – Bhadra Mohit  
**Kapsam:** 20 ülke, 2010–2019 dönemi  
**Veri türü:** Ekonomik, sosyal ve çevresel göstergeler

Veri seti, ülkelerin ekonomik ve sosyal durumlarını çevresel göstergelerle birlikte incelemeye olanak sağlayacak şekilde hazırlanmıştır. Bu nedenle tek bir değişkene odaklanmak yerine ülkeler arasındaki farklılıkları ve yıllar içerisindeki değişimleri birlikte değerlendirmek mümkündür.

### Değişkenler

| Değişken | Açıklama |
|---|---|
| `Country` | Verinin ait olduğu ülke |
| `Year` | Gözlemin ait olduğu yıl (2010–2019) |
| `GDP (USD)` | Gayrisafi yurt içi hasıla, milyar ABD doları |
| `Population` | Nüfus, milyon kişi |
| `Life Expectancy` | Doğuşta beklenen yaşam süresi, yıl |
| `Unemployment Rate (%)` | İşsizlik oranı |
| `CO2 Emissions (metric tons per capita)` | Kişi başına CO₂ emisyonu |
| `Access to Electricity (%)` | Elektriğe erişimi olan nüfusun oranı |

**Kaynak:** [Kaggle – World Bank Dataset](https://www.kaggle.com/datasets/bhadramohit/world-bank-dataset)


## 2. Analizde Kullanılan Sorular

Bu notebook boyunca aşağıdaki sorulara cevap aranmıştır:

1. GDP, nüfus ve kişi başına CO₂ emisyonlarının dağılımı nasıldır?
2. Ülkelerin işsizlik oranları yıllar içerisinde nasıl değişmektedir?
3. Yaşam beklentisinde ülkeler arasında ve yıllara göre nasıl farklılıklar vardır?
4. GDP ve nüfus yıllar içerisinde nasıl değişmektedir?
5. Elektriğe erişim oranları ülkeler arasında nasıl farklılaşmaktadır?
6. Ekonomik, sosyal ve çevresel değişkenler arasında hangi ilişkiler dikkat çekmektedir?


```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
sns.set_theme(style='whitegrid')
```

```python
# Veri setini yükleme (Klasör yapısına uygun olarak)
world_bank_data = pd.read_csv('../data/world_bank_dataset.csv')

print(f"Veri seti boyutu: {world_bank_data.shape[0]} satır, {world_bank_data.shape[1]} sütun")
world_bank_data.head()
```

## 3. Veri Setine İlk Bakış

Analize başlamadan önce veri setinin boyutu, değişken türleri ve eksik değerleri kontrol edilecektir. Bu adım, sonraki analizlerde kullanılacak verinin yapısını anlamak açısından önemlidir.


```python
# Değişkenler ve veri tipleri
world_bank_data.info()
```

```python
# Eksik değer kontrolü
missing_values = world_bank_data.isnull().sum().sort_values(ascending=False)
missing_values
```

## 4. Değişkenlerin Dağılımı

İlk görselleştirmede GDP, nüfus ve kişi başına CO₂ emisyonu değişkenlerinin dağılımlarına bakılmıştır. Böylece ülkeler arasındaki değer farklılıklarının genel görünümü incelenmektedir.


```python
columns_to_plot = [
    'GDP (USD)',
    'Population',
    'CO2 Emissions (metric tons per capita)'
]

for col in columns_to_plot:
    plt.figure(figsize=(8, 6))
    sns.histplot(world_bank_data[col], kde=True, bins=30)
    plt.title(f'{col} Dağılımı')
    plt.xlabel(col)
    plt.ylabel('Frekans')
    plt.tight_layout()
    plt.show()
```

### Yorum

Bu grafikler, GDP, nüfus ve kişi başına CO₂ emisyonlarının veri setindeki dağılımını göstermektedir. Dağılımın genişliği ve yoğunlaştığı bölgeler, ülkeler arasındaki değer farklılıkları hakkında ilk fikir vermektedir. Özellikle bazı değişkenlerde gözlenen yüksek değerlerin dağılımın sağ tarafında toplanması, ülkeler arasında belirgin ölçek farkları olabileceğini düşündürür. Bu nedenle sonraki analizlerde ortalama değerlerin yanında ülke ve yıl bazındaki değişimlerin de incelenmesi önemlidir.

## 5. Yıllara Göre İşsizlik Oranı

Bu bölümde ülkelerin işsizlik oranlarının 2010–2019 dönemindeki değişimi karşılaştırılmaktadır. Amaç, ülkeler arasındaki işsizlik seviyelerini ve zaman içerisindeki hareketleri görsel olarak incelemektir.


```python
avg_unemployment = (
    world_bank_data
    .groupby(['Year', 'Country'])['Unemployment Rate (%)']
    .mean()
    .reset_index()
)

plt.figure(figsize=(14, 8))
sns.barplot(
    data=avg_unemployment,
    x='Year',
    y='Unemployment Rate (%)',
    hue='Country',
    errorbar=None
)
plt.title('Yıllara Göre Ülkeler Arası İşsizlik Oranı', fontsize=16)
plt.xlabel('Yıl', fontsize=12)
plt.ylabel('İşsizlik Oranı (%)', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Ülke', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Yorum

İşsizlik oranları ülke ve yıl bazında karşılaştırıldığında, iş gücü piyasalarının zaman içerisindeki farklı hareketleri görülebilir. Grafikte çizgilerin/barların yıllar arasında değişmesi, ülkelerin işsizlik oranlarının dönem boyunca aynı seviyede kalmadığını; daha yatay bir görünüm ise daha sınırlı değişimi ifade eder. Ülkeler arasındaki farkların büyüklüğü değerlendirilirken tek bir yıla değil, 2010–2019 döneminin tamamına bakmak daha sağlıklı bir yorum yapılmasını sağlar.

## 6. Yaşam Beklentisinin Değişimi

Yaşam beklentisi, ülkelerin sosyal ve sağlık göstergelerini değerlendirmek için kullanılan önemli değişkenlerden biridir. Aşağıdaki grafik, ülkelerin yıllara göre yaşam beklentilerini karşılaştırmaktadır.


```python
plt.figure(figsize=(14, 8))
sns.barplot(
    data=world_bank_data,
    x='Year',
    y='Life Expectancy',
    hue='Country',
    errorbar=None
)
plt.title('Yıllara Göre Ülkeler Arası Ortalama Yaşam Süresi', fontsize=16)
plt.xlabel('Yıl', fontsize=12)
plt.ylabel('Yaşam Süresi (Yıl)', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Ülke', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Yorum

Yaşam beklentisi grafiği, ülkeler arasındaki sosyal göstergelerin farklılığını ve dönem içerisindeki değişimleri karşılaştırmayı sağlar. Bir ülkenin değerlerinin yıllar boyunca birbirine yakın seyretmesi daha istikrarlı bir görünüm sunarken, belirgin yükseliş veya düşüşler zaman içerisindeki değişime işaret eder. Bu göstergeyi GDP, elektrik erişimi ve işsizlik gibi diğer değişkenlerle birlikte değerlendirmek daha kapsamlı bir ülke karşılaştırması yapılmasına yardımcı olur.

## 7. GDP ve Nüfusun Yıllara Göre Değişimi

Heatmap kullanılarak her ülkenin yıllara göre GDP ve nüfus değerleri birlikte incelenmiştir. Renk yoğunluğu, ülkeler ve yıllar arasındaki değer farklarını daha kolay görmeyi sağlar.


```python
pivot_data_gdp = world_bank_data.pivot_table(
    index='Country',
    columns='Year',
    values='GDP (USD)',
    aggfunc='mean'
)

pivot_data_population = world_bank_data.pivot_table(
    index='Country',
    columns='Year',
    values='Population',
    aggfunc='mean'
)

fig, ax = plt.subplots(1, 2, figsize=(18, 8))

sns.heatmap(
    pivot_data_gdp,
    cmap='YlGnBu',
    annot=True,
    fmt='.1e',
    linewidths=0.5,
    ax=ax[0]
)
ax[0].set_title('Yıllara Göre Ülkeler Arası GDP (USD) Değişimi')
ax[0].set_xlabel('Yıl')
ax[0].set_ylabel('Ülke')

sns.heatmap(
    pivot_data_population,
    cmap='YlOrRd',
    annot=True,
    fmt='.1e',
    linewidths=0.5,
    ax=ax[1]
)
ax[1].set_title('Yıllara Göre Ülkeler Arası Nüfus Değişimi')
ax[1].set_xlabel('Yıl')
ax[1].set_ylabel('Ülke')

plt.tight_layout()
plt.show()
```

### Yorum

Heatmap'ler, ülkeler ve yıllar arasındaki GDP ve nüfus farklılıklarını tek bakışta karşılaştırmayı kolaylaştırmaktadır. Renk yoğunluklarının değişmesi, ülkelerin birbirinden farklı değer aralıklarına sahip olduğunu görmeyi sağlar. GDP ve nüfusun birlikte incelenmesi, ekonomik büyüklük ile nüfus ölçeğinin aynı yönde hareket edip etmediğine dair araştırma soruları oluşturmak için de kullanılabilir.

## 8. Elektriğe Erişim Oranı

Elektriğe erişim oranı, altyapı ve yaşam koşullarını değerlendirmede kullanılabilecek göstergelerden biridir. Bu bölümde ülkelerin yıllara göre elektrik erişim oranları karşılaştırılmıştır.


```python
plt.figure(figsize=(14, 8))
sns.barplot(
    data=world_bank_data,
    x='Year',
    y='Access to Electricity (%)',
    hue='Country',
    errorbar=None
)
plt.title('Yıllara Göre Ülkeler Arası Elektrik Erişim Oranı', fontsize=16)
plt.xlabel('Yıl', fontsize=12)
plt.ylabel('Elektrik Erişim Oranı (%)', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Ülke', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Yorum

Elektriğe erişim oranı, ülkelerin altyapı koşullarını karşılaştırmak açısından önemli bir göstergedir. Yıllar boyunca değerlerin yükselmesi erişimde iyileşmeye, benzer seviyelerde kalması ise daha istikrarlı bir görünüme işaret eder. Ülkeler arasındaki farkların zaman içerisinde kapanıp kapanmadığı da bu grafik üzerinden takip edilebilir.

## 9. Korelasyon Analizi

Son olarak sayısal değişkenler arasındaki doğrusal ilişkileri görmek amacıyla korelasyon matrisi oluşturulmuştur.

Korelasyon değerleri **nedensellik anlamına gelmez**. Bu bölüm yalnızca değişkenlerin birlikte nasıl hareket ettiğine dair genel bir fikir vermektedir.


```python
numeric_data = world_bank_data.select_dtypes(include=['float64', 'int64'])
correlation_matrix = numeric_data.corr()

plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)
plt.title('Korelasyon Matrisi')
plt.tight_layout()
plt.show()
```

### Yorum

Korelasyon matrisi, sayısal değişkenlerin birlikte hareket etme düzeylerini karşılaştırmayı sağlar. Pozitif korelasyon iki değişkenin aynı yönde, negatif korelasyon ise zıt yönde hareket etme eğilimini gösterir; değerin sıfıra yaklaşması doğrusal ilişkinin daha zayıf olduğunu ifade eder. Ancak korelasyonun tek başına nedensellik göstermediği unutulmamalıdır. Bu nedenle matriste dikkat çeken ilişkiler, ileride daha ayrıntılı istatistiksel analizlerle incelenebilir.

# 10. Key Findings / Bulgular

Aşağıdaki bölüm, grafiklere bakarak manuel olarak yorum yapmak yerine veri setindeki değerleri kullanarak bazı temel bulguları otomatik olarak özetler.

> Buradaki sonuçlar bu veri setinin kapsamıyla sınırlıdır. Özellikle veri seti, Kaggle açıklamasına göre gerçek Dünya Bankası verilerinin birebir arşivi değil; World Bank göstergelerini örnekleyen/simüle eden bir veri setidir.


```python
# Temel bulguları otomatik olarak üretme

# GDP ortalaması en yüksek 5 ülke
top_gdp_countries = (
    world_bank_data.groupby('Country')['GDP (USD)']
    .mean()
    .sort_values(ascending=False)
    .head(5)
)

# Ortalama yaşam beklentisi en yüksek 5 ülke
top_life_expectancy = (
    world_bank_data.groupby('Country')['Life Expectancy']
    .mean()
    .sort_values(ascending=False)
    .head(5)
)

# Ortalama işsizlik oranı en yüksek 5 ülke
top_unemployment = (
    world_bank_data.groupby('Country')['Unemployment Rate (%)']
    .mean()
    .sort_values(ascending=False)
    .head(5)
)

# Elektriğe erişim ortalaması en yüksek 5 ülke
top_electricity_access = (
    world_bank_data.groupby('Country')['Access to Electricity (%)']
    .mean()
    .sort_values(ascending=False)
    .head(5)
)

print("Ortalama GDP değeri en yüksek 5 ülke:")
display(top_gdp_countries)

print("\nOrtalama yaşam beklentisi en yüksek 5 ülke:")
display(top_life_expectancy)

print("\nOrtalama işsizlik oranı en yüksek 5 ülke:")
display(top_unemployment)

print("\nOrtalama elektriğe erişim oranı en yüksek 5 ülke:")
display(top_electricity_access)
```

```python
# Dönem içerisindeki değişimleri özetleme

yearly_means = world_bank_data.groupby('Year')[
    ['GDP (USD)', 'Population', 'Life Expectancy',
     'Unemployment Rate (%)',
     'CO2 Emissions (metric tons per capita)',
     'Access to Electricity (%)']
].mean()

print("Yıllara göre ortalama göstergeler:")
display(yearly_means.round(2))
```

### Bulguları Nasıl Okumalıyız?

- **GDP:** Ülkelerin ekonomik büyüklükleri arasında belirgin farklılıklar olup olmadığı incelenebilir.
- **Nüfus:** Ülkelerin nüfus yapılarındaki değişimler karşılaştırılabilir.
- **Yaşam beklentisi:** Ülkeler arasındaki sosyal göstergelerin farklılıkları görülebilir.
- **İşsizlik:** Ülkelerin iş gücü piyasalarındaki farklılıklar yıllar bazında incelenebilir.
- **CO₂ emisyonu:** Ekonomik ve çevresel göstergelerin birlikte nasıl hareket ettiği korelasyon analiziyle araştırılabilir.
- **Elektriğe erişim:** Altyapı göstergesinin ülkeler ve yıllar arasındaki değişimi değerlendirilebilir.
